```python
from scipy.optimize import curve_fit

# Data
kappa_plot = np.array(kappa_b_mult_list) * CircuitParams().kappa_b / (2 * np.pi) * 1e3
g2_list_threshold = np.array([
    jv(1, epsilon_p) * circuit.E_J * circuit.phi_a**2 * circuit.phi_b
    for epsilon_p in period_doubling_threshold
])

g2_plot = g2_list_threshold / (2 * np.pi) * 1e3

# Exponential model
def exp_fit(kappa, alpha, beta, gamma):
    return alpha + beta * np.exp(gamma * kappa)

# Fit
p0 = [g2_plot.min(), g2_plot.max() - g2_plot.min(), 0.1]
popt, pcov = curve_fit(exp_fit, kappa_plot, g2_plot, p0=p0)

alpha, beta, gamma = popt

# Smooth curve for plotting
kappa_fit = np.linspace(kappa_plot.min(), kappa_plot.max(), 300)
g2_fit = exp_fit(kappa_fit, alpha, beta, gamma)

# Plot
plt.figure()

plt.plot(
    kappa_plot,
    g2_plot,
    "o",
    label="Data",
)

plt.plot(
    kappa_fit,
    g2_fit,
    "--",
    label=rf"Fit: $\alpha+\beta e^{{\gamma \kappa_b}}$"
)

plt.xlabel(r"$\kappa_b/2\pi$" + " (MHz)")
plt.ylabel("Period-doubling threshold\n" + r"$g_2/2\pi$" + " (MHz)")

plt.legend()
plt.show()

print(f"alpha = {alpha:.6g}")
print(f"beta  = {beta:.6g}")
print(f"gamma = {gamma:.6g}")
```
